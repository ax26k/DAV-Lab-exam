```python
import pandas as pd
import os

# =========================
# LOAD DATA (from previous step)
# =========================

file_path = r"C:\Users\ub15-lab-004\Downloads\DAV data\final_cleaned_data.csv"

df = pd.read_csv(file_path)

print("Data loaded!")
print("Total rows:", len(df))


# =========================
# STEP 1: DEFINE REAL TRIP RULES
# =========================

# Convert datetime
df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])

# Calculate trip duration (minutes)
df['trip_duration'] = (
    df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']
).dt.total_seconds() / 60


# =========================
# STEP 2: BASIC CLEANING (REAL TRIPS)
# =========================

df_clean = df[
    (df['trip_distance'] > 0) &
    (df['fare_amount'] > 0) &
    (df['trip_duration'] > 1) &
    (df['trip_duration'] < 180)
]

print("\n--- BASIC CLEANING ---")
print("Valid trips:", len(df_clean))
print("Removed trips:", len(df) - len(df_clean))


# =========================
# STEP 3: STRICT CLEANING (FOR COMPARISON)
# =========================

df_strict = df[
    (df['trip_distance'] > 0.5) &
    (df['fare_amount'] > 2) &
    (df['trip_duration'] > 2) &
    (df['trip_duration'] < 120)
]

print("\n--- STRICT CLEANING ---")
print("Valid trips:", len(df_strict))
print("Removed trips:", len(df) - len(df_strict))


# =========================
# STEP 4: COMPARE IMPACT
# =========================

print("\n--- COMPARISON ---")
print("Original:", len(df))
print("Basic Clean:", len(df_clean))
print("Strict Clean:", len(df_strict))


# =========================
# STEP 5: SAVE CLEAN DATASETS
# =========================

folder_path = r"C:\Users\ub15-lab-004\Downloads\DAV data"

df_clean.to_csv(os.path.join(folder_path, "act2_basic_clean.csv"), index=False)
df_strict.to_csv(os.path.join(folder_path, "act2_strict_clean.csv"), index=False)

print("\nSaved:")
print("act2_basic_clean.csv")
print("act2_strict_clean.csv")

#“A real trip was defined using constraints on distance, fare, and duration. When stricter definitions were applied, a significant portion of trips were removed. This demonstrates that the system contains inconsistencies, and conclusions are highly sensitive to how ‘valid trips’ are defined."
```
